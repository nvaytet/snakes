```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from scipy.signal import convolve2d
from scipy.ndimage import median_filter
%matplotlib widget
import time
```

```python
nplayers = 2

plt.ioff()
cmap = mcolors.ListedColormap(['black'] + [plt.cm.tab10.colors[i] for i in range(nplayers)])
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(10, 6))
ax[0].set_axis_off()
ax[1].set_axis_off()
```

```python
def position_of_2_digit(a):
    n = np.asarray(a)
    # out = np.zeros_like(n)
    # sel = n > 0
    l = np.log10(n)
    d = n - 10**(l.astype(int))
    # print(l, d)
    # inds = d > 0
    # print(inds)
    # print(out[sel])
    # print(np.log10(d[inds]).astype(int))
    return np.log10(d).astype(int)
    # return out
```

```python
mult = np.arange(1, nplayers + 1).reshape(1, 1, -1)

class MultiplayerGameOfLife:
    def __init__(self, board_size=(160, 100), num_players=2, player_region_size=(10, 10)):
        self.board_size = board_size
        self.num_players = num_players
        self.player_region_size = player_region_size
        self.board = np.zeros(board_size, dtype=int)
        # self.board = np.full(board_size, 2, dtype=int)
        self.player_patterns = {}

    def set_player_pattern(self, player_id, pattern, top_left):
        self.player_patterns[player_id] = (pattern, top_left)

    def initialize_board(self):
        for player_id in range(self.num_players):
            pattern, top_left = self.player_patterns.get(player_id+1, (np.zeros(self.player_region_size), (0, 0)))
            bottom_right = (top_left[0] + pattern.shape[0], top_left[1] + pattern.shape[1])
            self.board[top_left[0]:bottom_right[0], top_left[1]:bottom_right[1]] = pattern * 10**(player_id)

        # self.image = ax[0].imshow(np.sum(self.board * mult, axis=2), cmap=cmap, interpolation='nearest', origin='lower')
        self.image = ax[0].imshow(np.log10(self.board), cmap=cmap, interpolation='nearest', origin='lower')
        # self.image = ax[0].imshow(self.board, cmap=cmap, interpolation='nearest', origin='lower')

    # def evolve_board(self):
    #     new_board = np.zeros_like(self.board)
    #     for i in range(self.board_size[0]):
    #         for j in range(self.board_size[1]):
    #             cell_state = self.board[i, j]
    #             num_neighbors = self.count_neighbors(i, j)
    #             new_cell_state = self.apply_rules(cell_state, num_neighbors)
    #             new_board[i, j] = new_cell_state
    #     self.board = new_board

    # def count_neighbors(self, i, j):
    #     count = 0
    #     for di in [-1, 0, 1]:
    #         for dj in [-1, 0, 1]:
    #             if di == 0 and dj == 0:
    #                 continue
    #             ni, nj = i + di, j + dj
    #             if 0 <= ni < self.board_size[0] and 0 <= nj < self.board_size[1]:
    #                 count += self.board[ni, nj]
    #     return count

    # def apply_rules(self, cell_state, num_neighbors):
    #     if cell_state == 1:  # Cell is alive
    #         return 1 if num_neighbors == 2 or num_neighbors == 3 else 0
    #     else:  # Cell is dead
    #         return 1 if num_neighbors == 3 else 0

    def display_board(self):
        # self.image.set_array(np.sum(self.board * mult, axis=2))
        self.image.set_array(self.board)
        fig.canvas.draw()

    def evolve_board(self):
        neighbor_filter = np.array([[1, 1, 1],
                                    [1, 0, 1],
                                    [1, 1, 1]], dtype=int)
        neighbor_count = np.zeros_like(self.board)
    
        # for player_id in range(self.num_players):
        #     player_cells = np.where(self.board == player_id, 1, 0).astype(np.uint8)
        #     player_cells = self.board[..., player_id]
        #     # print(np.pad(player_cells, 1).shape, neighbor_filter.shape)
        #     # print(player_id, neighbor_count.shape, np.pad(player_cells, 1).shape, np.convolve(np.pad(player_cells, 1), neighbor_filter, mode='valid').shape)
        #     # neighbor_count += np.convolve(np.pad(player_cells, 1), neighbor_filter, mode='valid')
        #     # print(player_cells.dtype, neighbor_filter.dtype)
        #     # neighbor_count = convolve2d(np.pad(player_cells, 1), neighbor_filter, mode='valid', boundary='wrap')

        cells = np.clip(self.board, 0, 1)
        neighbor_count = convolve2d(cells, neighbor_filter, mode='full', boundary='wrap')[1:-1, 1:-1]
        neighbor_median = median_filter(np.ma.masked_where(self.board == 0, self.board), size=3, mode='wrap')
        
        alive_mask = self.board > 0
        # alive_neighbor_count = neighbor_count[alive_mask]
        alive_neighbor_count = np.where(alive_mask, neighbor_count, 0)

        im = ax[1].imshow(neighbor_median, interpolation='nearest', origin='lower')
        plt.colorbar(im, ax=ax[1])
        fig.canvas.draw()
        # print('alive_mask', alive_mask.shape)
        # print('alive_neighbor_count', alive_neighbor_count.shape)
        # print('alive_neighbor_count', alive_neighbor_count)

        
        # Apply rules
        # self.board[alive_mask] = np.where((alive_neighbor_count == 2) | (alive_neighbor_count == 3), 
        #                                   self.board[alive_mask], 0)
        self.board = np.where((alive_neighbor_count == 2) | (alive_neighbor_count == 3), 
                                          self.board, 0)
        birth_mask = np.logical_and(~alive_mask, neighbor_count == 3)

        # im = ax[1].imshow(birth_mask, interpolation='nearest', origin='lower')
        # plt.colorbar(im, ax=ax[1])
        # fig.canvas.draw()

        # for player_id in range(1, self.num_players + 1):
        # self.board[birth_mask & (neighbor_count == 3)] = 1
        # print(position_of_2_digit(neighbor_sum))
        self.board = np.where(birth_mask & (neighbor_count == 3), neighbor_median, self.board)

    # def evolve_board(self):
    #     neighbor_filter = np.array([[1, 1, 1],
    #                                 [1, 0, 1],
    #                                 [1, 1, 1]])
    #     neighbor_count = np.zeros_like(self.board)
    
    #     for player_id in range(1, self.num_players + 1):
    #         player_cells = np.where(self.board == player_id, 1, 0)
    #         padded_player_cells = np.pad(player_cells, 1, mode='constant')
    #         conv_result = np.convolve(padded_player_cells.flatten(), 
    #                                   neighbor_filter.flatten(), 
    #                                   mode='valid')
    #         neighbor_count += conv_result.reshape((self.board_size[0] - 2, self.board_size[1] - 2))
    
    #     alive_mask = self.board > 0
    #     alive_neighbor_count = neighbor_count[alive_mask]
    
    #     # Apply rules
    #     self.board[alive_mask] = np.where((alive_neighbor_count == 2) | (alive_neighbor_count == 3), 
    #                                       self.board[alive_mask], 0)
    #     birth_mask = np.logical_and(~alive_mask, alive_neighbor_count == 3)
    #     for player_id in range(1, self.num_players + 1):
    #         self.board[birth_mask & (neighbor_count == 3 * player_id)] = player_id


```

```python
# Initialize the game
game = MultiplayerGameOfLife(board_size=(50, 50), num_players=2, player_region_size=(10, 10))

# Define player patterns
# pattern1 = np.array([[1, 1, 0], [1, 0, 1], [0, 1, 0]])
pattern1 = np.array([[0, 0, 0], [1, 1, 1], [0, 0, 0]])
pattern2 = np.array([[0, 1, 0], [0, 0, 1], [1, 1, 1]])

# Set player patterns
game.set_player_pattern(1, pattern1, (10, 10))
# game.set_player_pattern(1, pattern1, (13, 13))
game.set_player_pattern(2, pattern2, (30, 30))

# Initialize the game board
game.initialize_board()

# Display initial board state
game.display_board()
```

```python
fig.canvas
```

```python
# Evolve and display the board for 100 iterations
for _ in range(1):
    game.evolve_board()
    # game.display_board()
    time.sleep(0.1)
```

```python
import numpy as np

# Example array of numbers
numbers = np.array([120000, 123402, 230045, 345678])

# Find the position of the "2" digit in each number
position_of_2 = np.zeros_like(numbers, dtype=int)

# Find the position of the "2" digit using NumPy operations
for i, num in enumerate(numbers):
    position_of_2[i] = int(np.log10(num // 10**(int(np.log10(num)))) + 1)

print(position_of_2)
```
