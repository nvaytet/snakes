```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
import matplotlib.colors as mcolors
import time
from IPython import display

%matplotlib widget
plt.ioff()

# fig, ax = plt.subplots(
#     # figsize=(8, 8)
# )
# ax.set_axis_off()

nplayers = 2

cmap = mcolors.ListedColormap(['black'] + [plt.cm.tab10.colors[i] for i in range(nplayers)])
fig, ax = plt.subplots()
ax.set_axis_off()
```

```python
from numpy.lib.stride_tricks import as_strided

def grid_nD(arr):
    assert all(_len>2 for _len in arr.shape)
    
    nDims = len(arr.shape)
    newShape = [_len-2 for _len in arr.shape]
    newShape.extend([3] * nDims)
    
    newStrides = arr.strides + arr.strides
    return as_strided(arr, shape=newShape, strides=newStrides)


# index is number of neighbors alive
ruleOfLifeAlive = np.zeros(8+1, dtype=int)  # default all to dead
ruleOfLifeAlive[[2,3]] = 1 # alive stays alive <=> 2 or 3 neighbors
print("ruleOfLifeAlive:", ruleOfLifeAlive)

ruleOfLifeDead = np.zeros(8+1, dtype=int)   # default all to dead
ruleOfLifeDead[3] = 1      # dead switches to living <=> 3 neighbors
print(" ruleOfLifeDead:", ruleOfLifeDead)

# def imshow_helper(board, sleepSeconds=0.1):
#     ''' helper method to display a board for short time interval 
#         being lazy and using the MATLAB style API '''
#     plt.imshow(board, cmap="binary", interpolation="none")
#     plt.axis('off')
#     display.display(plt.gcf())
#     display.clear_output(wait=True)
#     time.sleep(sleepSeconds)
    
class GameOfLife(object):
    def __init__(self, board_size=(50,50)):
        full_size = tuple(i+2 for i in board_size)
        self.full = np.zeros(full_size, dtype=int)
        nd_slice = (slice(1, -1),) * len(board_size)
        # self.board = self.full[1:-1,1:-1,...] 
        self.board = self.full[nd_slice]
        self.ndims = len(self.board.shape)
        self.image = ax.imshow(self.board, origin='lower', norm=Normalize(0, nplayers + 1), cmap=cmap, interpolation='nearest')
        # fig.colorbar(self.image, ax=ax)
        # fig.canvas.draw_idle()

    def update_image(self):
        # print("updating")
        self.image.set_data(self.board)
        fig.canvas.draw()
        fig.canvas.flush_events()
        
    def run_board(self, N_ITERS = 10):
        # imshow_helper(self.board)
        for i in range(N_ITERS):
            print(i)
            neighborhoods = grid_nD(self.full)
            for n in range(nplayers):
                player_neigh = np.where(neighborhoods == n, 1, 0)
                print('max', player_neigh.max())
                # shape = (10,10) --> (-1, -2)
                sumOver = tuple(-(j+1) for j in range(self.ndims))
                neighborCt = np.sum(player_neigh, sumOver) - self.board
                self.board[:] = np.where(self.board, 
                                         ruleOfLifeAlive[neighborCt], 
                                         ruleOfLifeDead[neighborCt])
            # imshow_helper(self.board)
            # self.image.set_array(self.board)
            # fig.canvas.draw_idle()
            self.update_image()
            time.sleep(0.1)



gol = GameOfLife()
# gol.board[4:14, 4:14] = np.random.randint(0, 2, size=100).reshape(10, 10)
# gol.board[30:40, 30:40] = np.random.randint(0, 2, size=100).reshape(10, 10) * 2

gol.board[4:14, 4:14] = 1
gol.board[30:40, 30:40] = 1

gol.update_image()

fig.canvas
```
