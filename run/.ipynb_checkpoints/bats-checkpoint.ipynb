```python
import pythreejs as p3
import numpy as np

# Only plot one in 10 points to reduce the size of the output
N = 10000
pos_array = np.random.normal(scale=10, size=(N, 3)).astype('float32')
pos_array[:, 2] = 0

# Create a position buffer geometry
geometry = p3.BufferGeometry(
    attributes={'position': p3.BufferAttribute(array=pos_array)})
# Create a points material

tex = p3.ImageTexture(imageUri='bat.png')

material = p3.PointsMaterial(size=2, map=tex, transparent=True)
# Combine the geometry and material into a Points object
points = p3.Points(geometry=geometry, material=material)

# Create the scene and the renderer
view_width = 1000
view_height = 1000
camera = p3.PerspectiveCamera(position=[0, 0, 100], aspect=view_width/view_height)
# camera = p3.OrthographicCamera(-10, 10, -10, 10, -1, 300)
scene = p3.Scene(children=[points, camera], background="#DDDDDD")
controller = p3.OrbitControls(controlling=camera,
                              enableRotate=False,
)
renderer = p3.Renderer(camera=camera, scene=scene, controls=[controller],
                    width=view_width, height=view_height)
```

```python
vecs = np.random.uniform(-1, 1, size=(N, 2))
vecs /= np.linalg.norm(vecs, axis=1).reshape((-1, 1))
vecs

import time

def update(*args):
    for i in range(10000):
        pos_array[:, :2] += vecs * 0.1
        far = np.linalg.norm(pos_array, axis=1) > 100
        vecs[:, 0] = np.where(far, -vecs[:, 0], vecs[:, 0])
        vecs[:, 1] = np.where(far, -vecs[:, 1], vecs[:, 1])
        geometry.attributes['position'].array = pos_array.astype('float32')
        time.sleep(0.01)

import ipywidgets as ipw

b = ipw.Button(description='play')
b.on_click(update)

ipw.VBox([renderer, b])
```
