```python
import scipp as sc
import plopp as pp
%matplotlib widget
```

```python
import numpy as np

class DataGen:
    def __init__(self):
        self._N = 300
        self._xmin = -10.0
        self._xmax = 10.0
        self._ymin = -10.0
        self._ymax = 10.0
        self._data = sc.DataArray(data=sc.zeros(sizes={'y': self._N, 'x': self._N}),
                                 coords={'x': sc.linspace('x', self._xmin, self._xmax, self._N+1, unit='m'),
                                         'y': sc.linspace('y', self._ymin, self._ymax, self._N+1, unit='m')})

    def update(self):
        new_events = np.random.normal(scale=2, size=[100, 2])
        counts, _, _ = np.histogram2d(new_events[:, 0], new_events[:, 1],
                                      bins=(np.linspace(self._ymin, self._ymax, self._N+1),
                                            np.linspace(self._xmin, self._xmax, self._N+1)))
        self._data.values += counts

d = DataGen()
```

```python
def make_data():
    d.update()
    return d._data
```

```python
stream_node = pp.Node(make_data)
```

```python
fig = pp.figure2d(stream_node, norm='log')
```

```python
import asyncio

async def async_range(count):
    for i in range(count):
        yield(i)
        await asyncio.sleep(0.1)

async def my_stream_func():
    async for i in async_range(300):
        stream_node.notify_children('update')

streaming_task = asyncio.create_task(my_stream_func())
```
